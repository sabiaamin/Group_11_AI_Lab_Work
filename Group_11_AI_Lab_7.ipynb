# Aim

To implement Forward Chaining and Backward Chaining algorithms in Artificial Intelligence to perform logical reasoning and infer conclusions from a set of facts and rules.


# Algorithm

## 1. Forward Chaining
Forward Chaining is a data-driven inference technique that starts with known facts and repeatedly applies rules to derive new facts until the goal is reached or no more rules can be applied.

1. Initialize the set of known facts, rules, and the goal to be achieved.
2. Convert the initial facts into a set.
3. Examine each rule and check whether all its conditions are present in the known facts.
4. If all conditions are satisfied and the conclusion is not already known, add the conclusion to the facts.
5. Repeat the process until the goal is found or no new facts can be inferred.
6. If the goal is present in the facts, print that the goal has been achieved and return `True`.
7. Otherwise, print that the goal cannot be proved and return `False`.

## 2. Backward Chaining
Backward Chaining is a goal-driven inference technique that starts with a goal and works backward to determine whether the known facts and rules support it.

1. Initialize the goal, known facts, rules, and an empty set of visited goals.
2. Check whether the goal is already present in the known facts. If so, return `True`.
3. Check whether the goal has already been visited to avoid infinite loops.
4. Mark the current goal as visited.
5. Find the rules whose conclusions match the current goal.
6. For each matching rule, recursively try to prove all its conditions.
7. If all conditions are proved, conclude that the goal is proved and return `True`.
8. If no matching rule can prove the goal, print that the goal cannot be proved and return `False`.
9. Display whether the original goal was successfully proved.

## Forward Chaining

```python
def forward_chaining(facts, rules, goal):
    facts = set(facts)

    while True:
        new_fact_added = False

        for conditions, conclusion in rules:
            if set(conditions).issubset(facts):
                if conclusion not in facts:
                    facts.add(conclusion)
                    print("Inferred:", conclusion)
                    new_fact_added = True

        if goal in facts:
            print("Goal achieved:", goal)
            return True

        if not new_fact_added:
            break

    print("Goal cannot be proved:", goal)
    return False


# Initial facts
facts = ["A", "B"]

# Rules: (conditions, conclusion)
rules = [
    (["A", "B"], "C"),
    (["C"], "D"),
    (["D"], "E")
]

goal = "E"

print("Initial facts:", facts)
forward_chaining(facts, rules, goal)
```

## Backward Chaining

```python
def backward_chaining(goal, facts, rules, visited=None):
    if visited is None:
        visited = set()

    # If the goal is already a known fact
    if goal in facts:
        print("Fact found:", goal)
        return True

    # Avoid infinite loops
    if goal in visited:
        return False

    visited.add(goal)

    # Find rules that conclude the goal
    for conditions, conclusion in rules:
        if conclusion == goal:
            print("Trying to prove:", goal)

            # Every condition must be proved
            if all(
                backward_chaining(condition, facts, rules, visited.copy())
                for condition in conditions
            ):
                print("Proved:", goal)
                return True

    print("Cannot prove:", goal)
    return False


# Known facts
facts = {"A", "B"}

# Rules: (conditions, conclusion)
rules = [
    (["A", "B"], "C"),
    (["C"], "D"),
    (["D"], "E")
]

goal = "E"

print("Goal:", goal)

if backward_chaining(goal, facts, rules):
    print("Goal successfully proved.")
else:
    print("Goal could not be proved.")
```
