## Aim

To implement a Genetic Algorithm (GA) in Python to find the optimal solution to an optimization problem using the principles of natural selection and genetics.

# Algorithm: Genetic Algorithm

1. Import the `random` module.
2. Define the fitness function \(f(x) = x^2\), where \(x\) is an integer between 0 and 31.
3. Initialize a population of 4 random individuals.
4. Repeat the following steps for 10 generations:
   - **Selection:** Sort the population according to fitness in descending order and select the two fittest individuals as parents.
   - **Crossover:** Combine the bits of the two parents to generate two offspring.
   - **Mutation:** With a probability of 0.3, replace each offspring with a randomly generated individual.
   - **Population Update:** Form the new population using the two parents and two offspring.
   - **Fitness Evaluation:** Find and display the best individual and its fitness value for the current generation.
5. After all generations, identify the individual with the highest fitness.
6. Display the optimal solution and its maximum fitness value.

```python
# Aim: Find the maximum value of \(f(x)=x^2\), where \(x\) is an integer from 0 to 31, represented using a 5-bit chromosome.
import random

# Fitness function
def fitness(x):
    return x * x

# Initialize population
population = [random.randint(0, 31) for _ in range(4)]

for generation in range(10):
    # Selection: choose the two fittest individuals
    population.sort(key=fitness, reverse=True)
    parent1 = population[0]
    parent2 = population[1]

    # Crossover
    child1 = (parent1 & 24) | (parent2 & 7)
    child2 = (parent2 & 24) | (parent1 & 7)

    # Mutation
    if random.random() < 0.3:
        child1 = random.randint(0, 31)

    if random.random() < 0.3:
        child2 = random.randint(0, 31)

    # Create new population
    population = [parent1, parent2, child1, child2]

    best = max(population, key=fitness)
    print("Generation:", generation + 1,
          "Best x:", best, "Fitness:", fitness(best))

print("\nOptimal solution:", max(population, key=fitness))
print("Maximum fitness:", fitness(max(population, key=fitness)))
```
