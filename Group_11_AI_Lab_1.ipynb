3. Calculate predicted values using the current weights and bias.
4. Compute the prediction error and cost function.
5. Calculate gradients for both weights and the bias.
6. Update the weights and bias using the learning rate.

8. Display the final weights and bias.

## 2. Depth First Search (DFS)

1. Initialize a stack with the starting node.
2. Create an empty set to store visited nodes.
3. Pop a node from the stack.
4. If the node has already been visited, skip it.
5. Mark the node as visited and check whether it is the goal.
6. If the goal is found, display a success message and stop.
7. Otherwise, push its unvisited neighbours onto the stack.
8. Repeat until the goal is found or the stack becomes empty.
9. If the stack is empty and the goal is not found, display a failure message.

## 3. Breadth First Search (BFS)

1. Initialize a queue with the starting node.
2. Create an empty set to store visited nodes.
3. Remove a node from the front of the queue.
4. If the node has already been visited, skip it.
5. Mark the node as visited and check whether it is the goal.
6. If the goal is found, display a success message and stop.
7. Otherwise, add its unvisited neighbours to the queue.
8. Repeat until the goal is found or the queue becomes empty.
9. If the queue is empty and the goal is not found, display a failure message.

## Linear Regression

```python
import numpy as np
```

```python
X=np.array([
    [1000, 2],
    [1500, 3],
    [2000, 4],
    [2500, 4],
    [3000, 5]
],dtype=float)
```

```python
y=np.array([200, 300, 400, 500, 600], dtype=float)
```

```python
m=len(y)
w1=0
w2=0
b=0
alpha=0.0000001
iterations=10001
```

```python
for i in range(iterations):
    y_pred=w1*X[:,0]+w2*X[:,1]+b
    error=y_pred-y
    cost=np.sum(error**2)/(2*m)
    dw1=np.sum(error*X[:,0])/m
    dw2=np.sum(error*X[:,1])/m
    db=np.sum(error*1)/m

    w1=w1-alpha*dw1
    w2=w2-alpha*dw2
    b=b-alpha*db

    if i% 1000==0:
        print(f"iteration {i} cost {cost}")
```

```python

print("\nFinal Parameters:")
print("w1 =", w1)
print("w2 =", w2)
print("b =", b)
```

## Python Implementation of DFS

```python



def dfs(graph, start, goal):

    stack = [start]
    visited = set()

    while stack:

        node = stack.pop()

        if node in visited:
            continue

        print("Visiting:", node)
        visited.add(node)

        if node == goal:
            print("Goal found!")
            return

        for neighbour in reversed(graph[node]):
            if neighbour not in visited:
                stack.append(neighbour)

    print("Goal not found")


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

dfs(graph, 'A', 'F')
```

## Breadth First Search (BFS)

```python


from collections import deque

def bfs(graph, start, goal):

    queue = deque([start])
    visited = set()

    while queue:

        node = queue.popleft()

        if node in visited:
            continue

        print("Visiting:", node)
        visited.add(node)

        if node == goal:
            print("Goal found!")
            return

        for neighbour in graph[node]:
            if neighbour not in visited:
                queue.append(neighbour)

    print("Goal not found")


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

bfs(graph, 'A', 'F')
```
