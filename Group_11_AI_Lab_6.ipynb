### Objectives

Apply an activation function.

Update weights and bias using the perceptron learning rule.

Classify input data into two classes.

# Algorithm: Single-Layer Neural Network


1. Import the NumPy library.
2. Initialize the input dataset `X` with four input combinations for the AND gate.
3. Initialize the target output `y` as `[0, 0, 0, 1]`.
4. Initialize the weights and bias to zero.
5. Set the learning rate to `0.1` and the number of epochs to `10`.
6. Define the step activation function:
   - If the weighted sum is greater than or equal to zero, return `1`.
   - Otherwise, return `0`.
7. Repeat the training process for the specified number of epochs.
8. For each input sample:
   - Calculate the weighted sum using the input, weights, and bias.
   - Apply the activation function to predict the output.
   - Calculate the error as the difference between the actual and predicted outputs.
   - Update the weights using the perceptron learning rule.
   - Update the bias using the learning rate and error.
9. Display the updated weights and bias after each epoch.
10. Test the trained neural network on all input samples.
11. Display each input, its actual output, and its predicted output.

```python
import numpy as np

# Input data
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Target output for AND gate
y = np.array([0, 0, 0, 1])

# Initialize weights and bias
weights = np.zeros(2)
bias = 0

# Learning rate
learning_rate = 0.1

# Number of epochs
epochs = 10


# Step activation function
def activation(z):
    if z >= 0:
        return 1
    else:
        return 0


# Training the neural network
for epoch in range(epochs):

    print("Epoch:", epoch + 1)

    for i in range(len(X)):

        # Calculate weighted sum
        z = np.dot(X[i], weights) + bias

        # Predict output
        prediction = activation(z)

        # Calculate error
        error = y[i] - prediction

        # Update weights
        weights = weights + learning_rate * error * X[i]

        # Update bias
        bias = bias + learning_rate * error

    print("Weights:", weights)
    print("Bias:", bias)


# Testing
print("\nFinal Predictions:")

for i in range(len(X)):

    z = np.dot(X[i], weights) + bias

    prediction = activation(z)

    print(
        "Input:", X[i],
        "Actual:", y[i],
        "Predicted:", prediction
    )
```
