## Part A: 8-Puzzle Problem Using BFS

### Algorithm

1. Represent each puzzle configuration as a tuple containing 9 elements, where `0` represents the blank tile.
2. Define the initial state and the goal state.
3. Place the initial state and its path in a queue.
4. Remove the front state from the queue.
5. If the current state matches the goal state, return the solution path.
6. Generate neighbouring states by moving the blank tile up, down, left, or right whenever the move is valid.
7. Add each unvisited neighbouring state to the queue and record its path.
8. Repeat until the goal state is found or the queue becomes empty.
9. Display the solution path and the number of moves, or report that no solution was found.

```python
from collections import deque

# Initial and goal states
initial = (1, 2, 3,
           4, 0, 6,
           7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_zero = new_row * 3 + new_col

            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(tuple(new_state))

    return neighbors


def bfs(start, goal):
    queue = deque([(start, [start])])
    visited = {start}

    while queue:
        state, path = queue.popleft()

        if state == goal:
            return path

        for next_state in get_neighbors(state):
            if next_state not in visited:
                visited.add(next_state)
                queue.append((next_state, path + [next_state]))

    return None


solution = bfs(initial, goal)

if solution:
    print("Solution found!")
    print("Number of moves:", len(solution) - 1)

    for step, state in enumerate(solution):
        print("Step", step)
        print_puzzle(state)
else:
    print("No solution found.")
```

## Part B: 8-Queens Problem Using Backtracking

### Algorithm

1. Initialize an empty chessboard with 8 rows and 8 columns.
2. Place one queen in each row, trying columns one by one.
3. Check whether the chosen position is safe by verifying that no previously placed queen is in the same column or diagonal.
4. If the position is safe, place the queen and recursively process the next row.
5. If no safe position leads to a solution, remove the previously placed queen and try the next column. This step is called **backtracking**.
6. Continue until queens have been placed in all 8 rows.
7. Display the chessboard if a solution is found; otherwise, report that no solution exists.

```python
N = 8

board = [-1] * N


def is_safe(row, col):

    for previous_row in range(row):

        previous_col = board[previous_row]

        # Same column
        if previous_col == col:
            return False

        # Same diagonal
        if abs(previous_row - row) == abs(previous_col - col):
            return False

    return True


def solve(row):

    if row == N:
        return True

    for col in range(N):

        if is_safe(row, col):

            board[row] = col

            if solve(row + 1):
                return True

            # Backtracking
            board[row] = -1

    return False


def print_board():

    for row in range(N):

        for col in range(N):

            if board[row] == col:
                print("Q", end=" ")
            else:
                print(".", end=" ")

        print()


if solve(0):
    print("Solution for 8-Queens:")
    print_board()
else:
    print("No solution exists.")
```

## Part C: Tic-Tac-Toe Using Minimax Search

### Algorithm

1. Represent the Tic-Tac-Toe board as a list of 9 positions.
2. Define all possible winning combinations and check whether either player has won or the game is a draw.
3. Use the **Minimax** algorithm to evaluate possible future game states:
   - Assign a score of `1` if computer player `O` wins.
   - Assign a score of `-1` if human player `X` wins.
   - Assign a score of `0` if the game ends in a draw.
4. For the computer's turn, try each available position and recursively evaluate the resulting board.
5. Choose the move with the highest score for `O`; during the human player's simulated turn, choose the move with the lowest score for `X`.
6. Ask the human player to enter a move and update the board.
7. Calculate and make the computer's best move using Minimax.
8. Repeat the turns until a player wins or the game ends in a draw.
9. Display the board and the final result.

```python
import math

board = [" "] * 9


def print_board():

    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()


def check_winner():

    winning_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_positions:

        if board[a] == board[b] == board[c] != " ":
            return board[a]

    if " " not in board:
        return "Draw"

    return None


def minimax(is_maximizing):

    result = check_winner()

    if result == "O":
        return 1

    if result == "X":
        return -1

    if result == "Draw":
        return 0

    if is_maximizing:

        best_score = -math.inf

        for i in range(9):

            if board[i] == " ":

                board[i] = "O"

                score = minimax(False)

                board[i] = " "

                best_score = max(best_score, score)

        return best_score

    else:

        best_score = math.inf

        for i in range(9):

            if board[i] == " ":

                board[i] = "X"

                score = minimax(True)

                board[i] = " "

                best_score = min(best_score, score)

        return best_score


def best_move():

    best_score = -math.inf
    move = None

    for i in range(9):

        if board[i] == " ":

            board[i] = "O"

            score = minimax(False)

            board[i] = " "

            if score > best_score:
                best_score = score
                move = i

    return move


print("TIC-TAC-TOE")
print("Positions are 0 to 8")

print_board()

while True:

    # Human move
    position = int(input("Enter your move (0-8): "))

    if board[position] != " ":
        print("Position already occupied!")
        continue

    board[position] = "X"

    print_board()

    result = check_winner()

    if result:
        print("Result:", result)
        break

    # Computer move
    computer_move = best_move()
    board[computer_move] = "O"

    print("Computer selected:", computer_move)

    print_board()

    result = check_winner()

    if result:
        print("Result:", result)
        break
```
