# Algorithm

## 1. Best First Search

1. Initialize a priority queue with the starting node and its heuristic value.
2. Create a visited set and a parent dictionary.
3. Remove the node with the lowest heuristic value from the priority queue.
4. If the node has already been visited, skip it.
5. Mark the node as visited and check whether it is the goal node.
6. If the goal is found, reconstruct and display the path.
7. Otherwise, add the unvisited neighbouring nodes to the priority queue based on their heuristic values.
8. Repeat until the goal is found or the priority queue becomes empty.

## 2. A* (A-Star) Search

1. Initialize a priority queue with the starting node.
2. Set the cost of reaching the starting node to zero.
3. Maintain a parent dictionary to reconstruct the path.
4. Calculate the evaluation function using f(n) = g(n) + h(n), where g(n) is the path cost and h(n) is the heuristic value.
5. Select the node with the lowest evaluation cost.
6. If the goal node is reached, reconstruct and display the path.
7. Otherwise, calculate the costs of its neighbouring nodes and update their path costs and parents when a better path is found.
8. Repeat until the goal is reached or the priority queue becomes empty.

## 3. AO* (AO-Star) Search

1. Start with the initial node and its heuristic value.
2. Check whether the node is a terminal node. If so, mark it as solved.
3. Examine the available child-node groups or alternatives.
4. Calculate the estimated cost of each alternative using the heuristic values of its child nodes.
5. Select the alternative with the minimum estimated cost.
6. Update the heuristic value of the current node.
7. Expand the selected child nodes and continue the search.
8. Mark nodes as solved when their required child nodes are solved.
9. Continue until the solution is obtained.

## 4. Hill Climbing Algorithm

1. Initialize the current state with the starting state.
2. Evaluate the current state's value.
3. Identify the neighbouring states.
4. Select the neighbour with the highest value.
5. Compare the selected neighbour's value with the current state's value.
6. If the neighbour has a higher value, move to that state.
7. Repeat the process until no better neighbour is available.
8. Display the final state and its value.

## Best First Search


```python
import heapq

def best_first_search(graph, heuristic, start, goal):

    priority_queue = []
    heapq.heappush(priority_queue, (heuristic[start], start))

    visited = set()
    parent = {start: None}

    while priority_queue:

        h, current = heapq.heappop(priority_queue)

        if current in visited:
            continue

        visited.add(current)

        print("Expanding:", current)

        if current == goal:
            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            return path[::-1]

        for neighbor in graph[current]:

            if neighbor not in visited:
                parent[neighbor] = current
                heapq.heappush(
                    priority_queue,
                    (heuristic[neighbor], neighbor)
                )

    return None


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': ['G'],
    'G': []
}

heuristic = {
    'A': 6,
    'B': 4,
    'C': 3,
    'D': 5,
    'E': 2,
    'F': 1,
    'G': 0
}

path = best_first_search(graph, heuristic, 'A', 'G')

print("Path:", path)
```

## A* search

```python
import heapq

def a_star(graph, heuristic, start, goal):

    priority_queue = []

    heapq.heappush(priority_queue, (heuristic[start], 0, start))

    cost = {start: 0}
    parent = {start: None}

    while priority_queue:

        f, g, current = heapq.heappop(priority_queue)

        print("Expanding:", current)

        if current == goal:

            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            return path[::-1], g

        for neighbor, edge_cost in graph[current]:

            new_cost = g + edge_cost

            if neighbor not in cost or new_cost < cost[neighbor]:

                cost[neighbor] = new_cost
                parent[neighbor] = current

                f_value = new_cost + heuristic[neighbor]

                heapq.heappush(
                    priority_queue,
                    (f_value, new_cost, neighbor)
                )

    return None


graph = {
    'A': [('B', 2), ('C', 4)],
    'B': [('D', 3), ('E', 2)],
    'C': [('F', 3)],
    'D': [],
    'E': [('G', 3)],
    'F': [('G', 1)],
    'G': []
}

heuristic = {
    'A': 6,
    'B': 4,
    'C': 3,
    'D': 5,
    'E': 2,
    'F': 1,
    'G': 0
}

result = a_star(graph, heuristic, 'A', 'G')

print("Path:", result[0])
print("Cost:", result[1])
```

## A0* Search


```python
def ao_star(graph, heuristic, node, solved):

    if node not in graph or len(graph[node]) == 0:
        solved.add(node)
        return heuristic[node]

    best_cost = float('inf')
    best_option = None

    for option in graph[node]:

        cost = 0

        for child in option:
            cost += heuristic[child]

        if cost < best_cost:
            best_cost = cost
            best_option = option

    heuristic[node] = best_cost

    print("Expanding:", node)
    print("Selected:", best_option)

    for child in best_option:
        if child not in solved:
            ao_star(graph, heuristic, child, solved)

    solved.add(node)

    return heuristic[node]


graph = {
    'A': [['B'], ['C', 'D']],
    'B': [],
    'C': [],
    'D': []
}

heuristic = {
    'A': 10,
    'B': 4,
    'C': 2,
    'D': 3
}

solved = set()

cost = ao_star(graph, heuristic, 'A', solved)

print("Final cost:", cost)
print("Solved nodes:", solved)
```

## Hill Climbing

```python
def hill_climbing(values, start):

    current = start

    while True:

        print("Current:", current,
              "Value:", values[current])

        neighbors = []

        if current - 1 in values:
            neighbors.append(current - 1)

        if current + 1 in values:
            neighbors.append(current + 1)

        if not neighbors:
            break

        best_neighbor = max(
            neighbors,
            key=lambda x: values[x]
        )

        if values[best_neighbor] <= values[current]:
            break

        current = best_neighbor

    return current


values = {
    0: 2,
    1: 5,
    2: 8,
    3: 12,
    4: 10,
    5: 7
}

result = hill_climbing(values, 0)

print("Final state:", result)
print("Maximum value:", values[result])
```
